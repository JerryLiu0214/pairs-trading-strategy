```python
# ===== Cell 1：检查需要的库有没有装好 =====

import sys                      # sys 是 Python 自带的库，可以查看 Python 自身的信息
print("Python 版本:", sys.version)   # 打印当前 Python 版本，确认环境

# 下面逐个尝试导入项目需要的库；如果某个没装，会提示我们去装
for name in ["pandas", "numpy", "matplotlib", "sklearn", "xgboost", "tensorflow"]:
    try:                                    # try：先“试着”执行下面的代码
        module = __import__(name)           # __import__(name)：按名字导入一个库
        print(f"✅ {name} 已安装，版本 {module.__version__}")   # f"..." 是格式化字符串，{} 里放变量
    except ImportError:                     # 如果导入失败（没装），就执行这里
        print(f"❌ {name} 没有安装")
```
